```python
session = fastf1.get_session(2021, 'Las Vegas', 'Race')
# print(session.name)
session.load()
```

```python
session.results.head()
```

```python
session.laps.head()
```

```python
laps = session. laps
all_telemetry = []
for driver in session. drivers:
    driver_laps = laps.pick_drivers(driver)
    for _, lap in driver_laps.iterlaps():
        try:
            tel = lap.get_telemetry()
            tel['Driver'] = driver
            tel['LapNumber'] = lap["LapNumber"]
            all_telemetry.append(tel)
        except:
            pass # skips laps with missing telemetry
telemetry_df = pd.concat(all_telemetry)
telemetry_df
```
